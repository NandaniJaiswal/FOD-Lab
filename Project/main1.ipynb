### Feature Engineering

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

df = pd.read_csv('Smartphone_Usage_Dataset.csv')
```

```python
df['Social_Media_Ratio'] = df['Social_Media_Hours'] / (df['Daily_Phone_Hours'] + 1e-5)
df['Total_Weekly_Hours'] = (df['Daily_Phone_Hours'] * 5) + (df['Weekend_Screen_Time_Hours'] * 2)
categorical_features = ['Gender', 'Occupation', 'Device_Type']
numerical_features = ['Age', 'Daily_Phone_Hours', 'Social_Media_Hours', 
                      'Work_Productivity_Score', 'Sleep_Hours', 'App_Usage_Count', 
                      'Caffeine_Intake_Cups', 'Weekend_Screen_Time_Hours',
                      'Social_Media_Ratio', 'Total_Weekly_Hours']
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(drop='first', sparse_output=False), categorical_features)
    ])
X = df.drop(columns=['User_ID', 'Stress_Level'])
y = df['Stress_Level']
```

```python

X_transformed = preprocessor.fit_transform(X)
cat_feature_names = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features)
transformed_columns = numerical_features + list(cat_feature_names)

df_transformed = pd.DataFrame(X_transformed, columns=transformed_columns)
df_transformed['Stress_Level'] = y.values 
df_transformed.to_csv('Transformed_Smartphone_Usage_Dataset.csv', index=False)

print("Transformation Complete. Shape:", df_transformed.shape)
print(df_transformed.head())
```

### Linear Regression

```python
import numpy as np
import pandas as pd

df = pd.read_csv("Smartphone_Usage_Dataset.csv")
y = df['Stress_Level'].values
features_df = df.select_dtypes(include=[np.number]).drop(columns=['Stress_Level'], errors='ignore')

print("Features: ", list(features_df.columns))
```

```python
X = features_df.values
print(X.shape) 
X
```

```python
df_numeric = df.select_dtypes(include=np.number)

target_column = 'Stress_Level'
X = df_numeric.drop(columns=[target_column])
y = df_numeric[target_column]

# Preprocessing: Imputing and Scaling
imputer = SimpleImputer(strategy='mean')
X_imputed = imputer.fit_transform(X)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

X_final = np.hstack([np.ones((X_scaled.shape[0], 1)), X_scaled])


X_train, X_test, y_train, y_test = train_test_split(X_final, y.values, test_size=0.2, random_state=42)

# Gradient Descent for Linear Regression
def gradient_descent(X, y, lr=0.01, max_iter=1000):
    theta = np.zeros(X.shape[1])
    m = X.shape[0]

    for i in range(max_iter):
        y_hat = X @ theta
        grad = (X.T @ (y_hat - y)) / m
        theta -= lr * grad

    return theta

# Train model
theta = gradient_descent(X_train, y_train, lr=0.01, max_iter=1000)
print("Learned parameters (theta):", theta)


def mse(y_pred, y_true):
    return ((y_pred - y_true)**2).mean()

# Training MSE
y_train_pred = X_train @ theta
mse_train = mse(y_train_pred, y_train)
print(f"MSE train: {mse_train:.4f}")

# Testing MSE
y_test_pred = X_test @ theta
mse_test = mse(y_test_pred, y_test)
print(f"MSE test: {mse_test:.4f}")

# Visualize Predictions
plt.figure(figsize=(6,6))
plt.scatter(y_test, y_test_pred, c='crimson', alpha=0.3)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'b--', linewidth=2)
plt.xlabel('True Stress Level')
plt.ylabel('Predicted Stress Level')
plt.title('Linear Regression: True vs Predicted')
plt.axis('equal')
plt.show()
```

```python
# Inference 
print('Sample value test -->', X_test[0] @ theta)
print('Actual value -->', y_test[0])
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler


def sigmoid(z):
    return 1 / (1 + np.exp(-z))


df = pd.read_csv('Smartphone_Usage_Dataset.csv')
target_data = df['Stress_Level'].values.reshape(-1, 1)


scaler = StandardScaler()
target_scaled = scaler.fit_transform(target_data)

target_sigmoid = sigmoid(target_scaled)


df['Stress_Level_sigmoid'] = target_sigmoid

print(df[['Stress_Level', 'Stress_Level_sigmoid']].head())
```

```python
def sigmoid(x, deriv=False):
    sig = 1 / (1 + np.exp(-x))
    if deriv:
        return sig * (1 - sig)
    return sig

stress = df["Stress_Level"].values.reshape(-1, 1)
scaler = StandardScaler()
stress_scaled = scaler.fit_transform(stress).flatten()

# Apply sigmoid + derivative
stress_sigmoid = sigmoid(stress_scaled)
stress_deriv = sigmoid(stress_scaled, deriv=True)

df["Stress_sigmoid"] = stress_sigmoid
df["Stress_sigmoid_derivative"] = stress_deriv

print(df[["Stress_Level", "Stress_sigmoid", "Stress_sigmoid_derivative"]].head())
```

### Prepare Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# 1. Load the dataset
df = pd.read_csv("Smartphone_Usage_Dataset.csv")

# 2. Convert Stress_Level to binary (above median = 1, below = 0)
median_stress = df["Stress_Level"].median()
df["Stress_binary"] = (df["Stress_Level"] > median_stress).astype(int)

# 3. Features and Target
# Selecting only numeric columns and dropping both target versions
X = df.select_dtypes(include=np.number).drop(columns=["Stress_Level", "Stress_binary"])
y = df["Stress_binary"].values

# 4. Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 5. Add intercept (Bias term)
X_scaled = np.hstack([np.ones((X_scaled.shape[0], 1)), X_scaled])

# 6. Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"y_train distribution: {np.bincount(y_train)}")
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_loss(y, h):
    epsilon = 1e-10
    h = np.clip(h, epsilon, 1 - epsilon)
    return -np.mean(y * np.log(h) + (1 - y) * np.log(1 - h))

# Gradient Descent Training
def train_logistic_regression(X, y, lr=0.01, iterations=1000):
    m, n = X.shape
    theta = np.zeros(n)
    losses = []

    for i in range(iterations):
        # Forward Pass
        z = X @ theta
        h = sigmoid(z)

        # Gradient Calculation
        gradient = (X.T @ (h - y)) / m
        
        # Update Rule
        theta -= lr * gradient

        # Track Progress
        loss = compute_loss(y, h)
        losses.append(loss)

        if i % 100 == 0:
            print(f"Iteration {i} | Loss: {loss:.4f}")

    return theta, losses

# Train the Model
theta, losses = train_logistic_regression(
    X_train, y_train, 
    lr=0.05,          # learning rate
    iterations=2000   # more iterations = better convergence
)
```

### Accuracy

```python

def predict(X, theta):
    h = sigmoid(X @ theta)
    return [1 if i >= 0.5 else 0 for i in h]

# Compute Accuracy 
y_pred = predict(X_test, theta)
accuracy = np.mean(y_pred == y_test)
print(f"Test Accuracy: {accuracy * 100:.2f}%")

# Inference on first test sample
print("Single Sample Inference ")
print("Actual Class: ", y_test[0])

# h_theta calculation (probability)
h_test_0 = sigmoid(X_test[0] @ theta)

# Final prediction (thresholded)
yhat = 1 if h_test_0 >= 0.5 else 0
print(f"Predicted Probability: {h_test_0:.4f}")
print(f"Predicted Class (yhat): {yhat}")
```

```python
from sklearn.metrics import accuracy_score

z_test = X_test @ theta
h_test = 1 / (1 + np.exp(-z_test))
yhat = (h_test >= 0.5).astype(int)

accuracy = accuracy_score(y_test, yhat)

print(f"Accuracy Score: {accuracy:.4f}")
print(f"Accuracy Percentage: {accuracy * 100:.2f}%")
```
